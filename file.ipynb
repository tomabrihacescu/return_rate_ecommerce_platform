```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

df =pd.read_csv("data/track_k_ecommerce_returns_train.csv")
print(df.head())

plt.scatter(df['price_eur'], df['return_rate_pct'],color ='blue')
plt.xlabel("price")
plt.ylabel("image quality")
plt.show()



plt.scatter(df['price_eur'], df['packaging_color_code'],color ='blue')
plt.xlabel("price")
plt.ylabel("color")
plt.show()

plt.scatter(df['price_eur'], df['shipping_days'],color ='blue')
plt.xlabel("price")
plt.ylabel("color")
plt.show()


import seaborn as sns
sns.pairplot(df.drop(columns=['size_variants_count', 'review_volume', 'size_mismatch_reports', 'holiday_campaign_flag', 'packaging_color_code']))
sns.heatmap(df.corr(), cmap="jet", vmin ='-1')
```
